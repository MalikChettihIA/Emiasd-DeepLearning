```python
from alphazero_lite_go import AlphaZeroLite, GoGame
from tensorflow import keras

# Charger l’ancien modèle
model = keras.models.load_model("mchettih.h5", compile=True)

# Réentraînement par self-play
az = AlphaZeroLite(model, GoGame, num_simulations=50)
az.train_from_selfplay(n_games=5, batch_size=64, epochs=3)

# Sauvegarder la nouvelle version
model.save("mchettih_v2.h5")
```

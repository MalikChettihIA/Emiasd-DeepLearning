```python
import numpy as np

PLANES = 5
MAX_LIB = 6

def move_to_coords(move):
    return divmod(move, 19)  # (row, col)

def parse_line(line):
    tokens = line.strip().split()
    komi = float(tokens[0])
    winner = tokens[1]
    nb_moves = int(tokens[2])
    moves = []
    for i in range(nb_moves):
        offset = 3 + i * 4
        inter = int(tokens[offset])
        color = int(tokens[offset + 1])
        val = float(tokens[offset + 2])
        points = float(tokens[offset + 3])
        moves.append((inter, color, val, points))
    return komi, winner, moves

def build_board_and_history(moves):
    board = [['.' for _ in range(19)] for _ in range(19)]
    history = []
    for move_idx, (inter, color, _, _) in enumerate(moves):
        if inter == 361:
            continue
        i, j = move_to_coords(inter)
        board[i][j] = 'B' if color == 1 else 'W'
        snap = [row.copy() for row in board]
        history.append(snap)
    return board, history

def encode_input(board, history, turn, komi):
    input_tensor = np.zeros((19, 19, 2 * PLANES + 5 + MAX_LIB + 10), dtype=np.float32)

    # Plan 0 : joueur courant
    input_tensor[:, :, 0] = 1.0 if turn == 'B' else 0.0

    # Plans 1-4 : mock données (échelles etc. non calculées ici)
    # Tu peux ajouter des heuristiques plus fines ici
    input_tensor[:, :, 1:5] = 0.0

    # Plans 5-10 : libertés simulées (au hasard)
    for i in range(19):
        for j in range(19):
            if board[i][j] != '.':
                libs = np.random.randint(1, MAX_LIB + 1)
                input_tensor[i, j, 4 + libs + 1] = 1.0

    # Plans 11-20 : historique
    for p in range(min(PLANES, len(history))):
        hist = history[-(p + 1)]
        for i in range(19):
            for j in range(19):
                if hist[i][j] == turn:
                    input_tensor[i, j, 11 + 2 * p] = 1.0
                elif hist[i][j] != '.':
                    input_tensor[i, j, 11 + 2 * p + 1] = 1.0

    # Plans 21-30 : komi one-hot
    komi_int = int(komi)
    if 0 <= komi_int < 10:
        input_tensor[:, :, 21 + komi_int] = 1.0

    return input_tensor

def from_line_to_tensor(line):
    komi, winner, moves = parse_line(line)
    board, history = build_board_and_history(moves)
    turn = 'B' if len(moves) % 2 == 0 else 'W'
    return encode_input(board, history, turn, komi)

```

```python
import numpy as np

def load_all_inputs(file_path):
    input_tensors = []
    total_lines = 0

    with open(file_path, "r") as f:
        nb_examples = int(f.readline())  # On lit et ignore le header
        for i, line in enumerate(f):
            try:
                input_tensor = from_line_to_tensor(line)
                input_tensors.append(input_tensor)
                total_lines += 1
            except Exception as e:
                print(f"[Erreur à la ligne {i+2}] {e}")
                continue

    print(f"\n✅ {total_lines} lignes lues avec succès.")
    return np.array(input_tensors, dtype=np.float32)

# === Utilisation ===
if __name__ == "__main__":
    all_inputs = load_all_inputs("games2022.data")
    print(f"Tenseur global : shape {all_inputs.shape}")# (19, 19, 31)
```
